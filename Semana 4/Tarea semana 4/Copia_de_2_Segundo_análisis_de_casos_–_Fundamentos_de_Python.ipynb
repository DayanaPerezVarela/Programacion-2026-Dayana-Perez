# Problema #1: Calculadora de propina inteligente

## Problema

Resolver la cuenta total a pagar con la propina incluida.

## Entrada

Solicitar el monto que va a pagar, y que porcentaje de propina va a dar
## Proceso

Multiplicar la cuenta por el porcentaje de propina divido entre 100
y sumar la propina y la cuenta

## Salida esperada

El monto de su propina es... pagas en total...

```python
#solicitud de datos
try:
  total_cuenta = float(input("Ingresa el monto de la cuenta:"))
except ValueError:
  print("Cuenta invalida, por favor intentar de nuevo e ingresar un número valido, mientras se asinga 1")
  total_cuenta = 1

try:
  porcent_pro = float(input("Ingresa el porcentaje de propina que quieres dar:"))
except ValueError:
  print("Intenta de nuevo, ingresa un número del 1 - 100, mientras se le asigna 10")
  porcent_pro = 10

#proceso
total_propina = total_cuenta * (porcent_pro/100)
monto_total = total_cuenta + total_propina

#salida
print(f"El monto a pagar de propina es: ₡{total_propina:.2f}")
print(f"El monto total a pagar es:₡{monto_total:.2f}")
```

# Problema #2: Detector de número par/impar

## Problema

Saber si un numero es par o impar

## Entrada

Solicitar un número.

## Proceso

dividir el número si no es un número entero es impar

## Salida esperada

Su número es impar o par.

```python
# entrada
try:
  numero = int(input("Ingresa un número:"))
except ValueError:
  print("Ingresa un número entero, se asigna 1")
  numero = 1

#proceso
if numero % 2 == 0 :
  print(f"El {numero} es un número par")
else:
  print(f"El {numero} es un  número impar")
```

# Problema #3: Clasificador de edad

## Problema

Clasificar si es un menor, adulto o adulto mayor.

## Entrada
Solicitar la edad de la persona

## Proceso

Comparar la edad para verificar si es mayor de 18, menor de 18 o mayor de 65 adulto mayor


## Salida esperada

-

```python
# Escriba la solución aquí
#Entrada
try:
  edad = int(input("Ingrese su edad actual:"))
except ValueError:
  print("Edad incorrecta:Ingresa un número, mientras se asignara 1")
  edad = 1

if edad < 18 :
  print("Su edad corresponde a un Menor de Edad")
elif edad >= 18 and edad <65:
  print("Su edad corresponde a un Adulto")
else :
  print("Su edad corresponde a un Adulto Mayor")
```

# Problema #4: Validador de compra con descuento

## Problema

Dar un descuento de 20% si el cliente paga más de 100000 y si es cliente frecuente, dar un descuento del 10% si es cliente frecuente o el monto es mayor a 80mil.

## Entrada

Solicitar el monto a pagar y solicitar si es cliente frecuente

## Proceso

Ver si aplica un descuento, si no, paga el total.

## Salida esperada

Su monto es de .... tiene un descuento por ser cliente frecuente de ... su monto con descuento aplicado es...

## Pruebas manuales
Cliente con monto mayor a 100000 y cliente frecuente obtiene un 20% de descuento = si se obtiene

cliente con monto menor a 80 000 pero si es cliente frecuente obtiene 10% de descuento = si se obtiene

cliente con monto mayor a 80 000 y no es cliente frecuente
obtiene 10% de descuento = si se obtiene

cliente con montos menores a 80 000 y no cliente frecuente no obtiene descuento == si se obtiene


```python
# Escriba la solución aquí
#Entrada
nombre = input("Ingresa tu nombre:")

try:
  monto_neto = float(input("Ingresa el monto de tu factura"))
except ValueError :
  print("Por favor ingresa el monto en números. Se asignara un 1")
  monto_neto = 1

cliente_fre = input("Eres cliente frecuente?(Si/No)").strip().lower() == "si"

#proceso
mensaje = ""
if monto_neto >= 100000 and cliente_fre:
  descuento = monto_neto * 0.20
  mensaje = "genial, obtienes un descuento del 20% por ser nuestro cliente frecuente"
elif cliente_fre or monto_neto >= 80000:
  descuento = monto_neto * 0.10
  mensaje = "te daremos un descuento del 10%"
else :
  descuento = 0
  mensaje = "lo sentimos, no tenemos descuento para usted, afiliate para disfrutar de más beneficios"

total = monto_neto - descuento

print(f"Hola {nombre}, {mensaje}, el monto total a pagar es: ₡{total:.2f}")
```
